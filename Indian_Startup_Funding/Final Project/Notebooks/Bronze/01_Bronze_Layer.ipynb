```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("IndianStartupFunding_Bronze") \
    .getOrCreate()

storage_account = "startupfundingstorage"

raw_path = f"abfss://raw@{storage_account}.dfs.core.windows.net/"
source_file = f"{raw_path}indian_startup_funding_2020_2025_sample.csv"

bronze_path = f"abfss://bronze@{storage_account}.dfs.core.windows.net/startup_funding"

df_raw = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "false") \
    .option("multiLine", "true") \
    .option("escape", '"') \
    .load(source_file)

print("Source Data:")
display(df_raw)

print("Schema:")
df_raw.printSchema()

print("Number of Rows:")
print(df_raw.count())

print("Number of Columns:")
print(len(df_raw.columns))

print("Column Names:")
print(df_raw.columns)

df_raw.write \
    .format("delta") \
    .mode("append") \
    .save(bronze_path)

print("Bronze Delta data written successfully.")

bronze_df = spark.read \
    .format("delta") \
    .load(bronze_path)

print("Bronze Row Count:")
print(bronze_df.count())

print("Bronze Schema:")
bronze_df.printSchema()

display(bronze_df)
```
