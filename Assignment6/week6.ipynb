```python
import pandas as pd
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Week6Assignment") \
    .getOrCreate()

# Read Excel using Pandas
pdf = pd.read_excel("source_dataset.xlsx")

# Convert Pandas DataFrame to Spark DataFrame
df = spark.createDataFrame(pdf)



df.filter(df.category == "Electronics") \
  .select("product_id", "price") \
  .show()

from pyspark.sql.functions import col

df = df.withColumnRenamed("old_name", "new_name")

df = df.withColumn(
    "price",
    col("price").cast("double")
)

df.printSchema()


df.filter(
    (df.status == "Completed") &
    (df.amount > 1000)
).show()


from pyspark.sql.functions import col

df = df.withColumn(
    "final_price",
    col("base_price") * 1.18
)

df.select(
    "product_id",
    "base_price",
    "final_price"
).show()





df.write.mode("overwrite").parquet("data/parquet_data")
parquet_df = spark.read.parquet("data/parquet_data")
filtered_df = parquet_df.filter(
    parquet_df.user_id.isNotNull()
)
filtered_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("output/result_csv")


df.filter(
    (df.region == "North") |
    (df.priority == "High")
).show()
```
